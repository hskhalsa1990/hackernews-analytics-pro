# HackerNews Analytics Pro — v13 Story Recommendations

Content-based discovery using TF-IDF word/bigram vectors, cosine similarity, and near-duplicate screening. This is lexical similarity, not a pretrained semantic model.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if not (ROOT / 'hn_analytics.py').exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
from hn_data_loader import load_stories
from hn_recommend_v13 import recommend_stories, find_near_duplicates
stories = load_stories(ROOT / 'demo_stories.jsonl')
stories.shape
```

```python
seed_index = 0
print('Seed:', stories.iloc[seed_index].get('title', ''))
recommendations = recommend_stories(stories, seed_index=seed_index, top_k=10, min_similarity=0.03)
recommendations
```

```python
duplicates = find_near_duplicates(stories, threshold=0.70)
duplicates.head(25)
```

```python
import matplotlib.pyplot as plt
if not recommendations.empty:
    plot = recommendations.sort_values('similarity')
    plt.figure(figsize=(10, 6))
    plt.barh(plot['title'].astype(str), plot['similarity'])
    plt.xlabel('Cosine similarity')
    plt.title('Stories most similar to selected story')
    plt.tight_layout()
    plt.show()
```

## Interpretation
- Similarity uses title plus optional text/description fields.
- The seed story is excluded from its own recommendations.
- Similarity is lexical overlap, so synonyms and context can be missed.
- Near-duplicate pairs are candidates for manual review, not automatically confirmed duplicates.
