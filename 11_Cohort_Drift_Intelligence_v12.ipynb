# v12 — Author, Domain & Drift Intelligence

This notebook profiles author/domain cohorts and screens chronological distribution shift. Metrics are descriptive, not causal. PSI thresholds are heuristic.


```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if not (ROOT / "hn_analytics.py").exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
import pandas as pd
from hn_data_loader import load_stories
from hn_profile_v12 import cohort_profiles, chronological_drift, export_v12_profile
raw = load_stories(ROOT / "demo_stories.jsonl")
print(raw.shape)
```

## 1. Author and domain cohorts

Minimum cohort size avoids over-interpreting one-off entities. Wilson intervals show sampling uncertainty for the top-decile rate, not causal uncertainty.

```python
authors, domains = cohort_profiles(raw, min_stories=3, top_n=30)
display(authors.head(15))
display(domains.head(15))
```

## 2. Chronological drift screen

The earlier segment defines PSI bins; the later segment is compared against it. PSI is a screening statistic, not a formal hypothesis test.

```python
drift, metadata = chronological_drift(raw, reference_fraction=0.5, bins=10)
display(drift)
metadata
```
