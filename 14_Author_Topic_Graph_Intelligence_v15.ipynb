# HackerNews Analytics Pro — v15 Author-Topic Graph Intelligence

This notebook constructs an author/topic bipartite graph from stories and exploratory topic assignments. Shared topics indicate overlap in observed content, **not collaboration or influence**.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if not (ROOT / 'hn_analytics.py').exists() and (ROOT.parent / 'src').exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
import pandas as pd
from hn_data_loader import load_stories
from hn_graph_v15 import build_author_topic_graph, export_v15_graph
stories = load_stories(ROOT / 'demo_stories.jsonl')
stories.head()
```

## Build topic assignments and graph metrics
The default TF-IDF representation is deterministic and works offline. You can select `auto` to use optional Sentence-Transformers when available.

```python
authors, topics, author_topic_edges, author_overlaps, graph, metadata = build_author_topic_graph(
    stories, n_clusters=min(5, max(2, len(stories))), backend='tfidf'
)
display(authors.head(20))
display(topics)
display(author_topic_edges.head(20))
metadata
```

## Explore the author projection
Weighted Jaccard measures similarity between authors' per-topic story-count profiles. It does not establish social ties.

```python
display(author_overlaps.head(30))
print('Graph nodes:', len(graph['nodes']))
print('Author-topic links:', len(graph['links']))
```

## Export reproducible artifacts

```python
result = export_v15_graph(stories, ROOT / 'reports', n_clusters=min(5, max(2, len(stories))), backend='tfidf')
result
```
