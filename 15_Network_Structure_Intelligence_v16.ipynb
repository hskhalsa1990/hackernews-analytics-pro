# Author Network Structure Intelligence — v16

This notebook ranks nodes in the observed author-topic graph using weighted PageRank and groups authors by thresholded topic-profile overlap. These are descriptive structural metrics, not measures of influence or collaboration.

```python
from pathlib import Path
import pandas as pd
from hn_data_loader import load_stories
from hn_network_v16 import build_v16_network

stories = load_stories("../demo_stories.jsonl")
authors, topics, communities, graph, metadata = build_v16_network(stories, n_clusters=5, backend="tfidf", similarity_threshold=0.15)
metadata
```

```python
display(authors.head(15))
display(topics)
display(communities)
```

```python
import matplotlib.pyplot as plt
authors.head(15).sort_values("pagerank").plot.barh(x="author", y="pagerank", legend=False, title="Top author nodes by weighted PageRank")
plt.xlabel("PageRank")
plt.tight_layout()
plt.show()
```

## Interpretation and limitations

PageRank measures prominence in this constructed bipartite network. Community IDs are connected components after applying the selected weighted-Jaccard threshold. A threshold change can merge or split groups. The graph is built from a bounded dataset and does not establish social relationships, influence, collaboration, or causation.
