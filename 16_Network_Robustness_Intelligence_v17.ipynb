# Network Robustness Intelligence — v17

How stable are the v16 PageRank rankings? This notebook bootstraps stories (topic labels held fixed), checks sensitivity to the damping factor and scores bridge nodes with betweenness. These are descriptive structural diagnostics, not influence or collaboration measures.

```python
from hn_data_loader import load_stories
from hn_robustness_v17 import build_v17_robustness

stories = load_stories("../demo_stories.jsonl")
authors, sensitivity, topics, summary = build_v17_robustness(stories, n_clusters=5, repeats=200, top_k=5)
summary
```

```python
display(authors.head(15))
display(sensitivity)
display(topics)
```

```python
import matplotlib.pyplot as plt
top = authors.head(10).iloc[::-1]
plt.errorbar(top.pagerank, top.author, xerr=[top.pagerank-top.ci_low, top.ci_high-top.pagerank], fmt="o")
plt.xlabel("PageRank (95% bootstrap interval)")
plt.tight_layout(); plt.show()
```

## Limitations

Bootstrap intervals reflect story-sampling noise only; topic labels are fixed from one clustering run, so clustering instability is not captured. Betweenness is unweighted. Small samples give wide intervals. Nothing here establishes influence, collaboration or causation.
